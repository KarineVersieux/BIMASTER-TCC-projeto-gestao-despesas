# CONFIGURAÇÃO (CLONAR PROJETO, CONFIGURAR AMBIENTE)

```python
# ------------------------------------------------------------
# 1. Clonar o projeto GitHub -> Colab
# chave github: GITHUB_OBRAS_TOKEN
# ------------------------------------------------------------

import subprocess
import os

repo_dir = "/content/BIMASTER-TCC-projeto-gestao-despesas"


repo_url = (
    "https://github.com/KarineVersieux/"
    "BIMASTER-TCC-projeto-gestao-despesas.git"
)

print(repo_url)

result = subprocess.run(
    ["git", "clone", repo_url, repo_dir],
    capture_output=True,
    text=True
)

if result.returncode == 0:
    print("Projeto clonado com sucesso!")
else:
    print("Erro ao clonar:")
    print(result.stderr)
```

```python
# ------------------------------------------------------------
# 2. Definir pasta do projeto
# ------------------------------------------------------------
%cd /content/BIMASTER-TCC-projeto-gestao-despesas
pwd = !pwd
print("Pasta do projeto:", pwd[0])
```

```python
# ------------------------------------------------------------
# 3. Carregar o conteúdo .env
# ------------------------------------------------------------

!pip install python-dotenv -q

from dotenv import load_dotenv
import os

# Carrega as variáveis do .env
load_dotenv("/content/.env", override=True)


# Modelos
print("SQL_LLM_MODEL:", os.getenv("SQL_LLM_MODEL"))
print("EXTRACT_LLM_MODEL:", os.getenv("EXTRACT_LLM_MODEL"))

# Projeto
repo_dir = "/content/BIMASTER-TCC-projeto-gestao-despesas"

print("Projeto existe:", os.path.exists(repo_dir))
```

```python
# ------------------------------------------------------------
# 4. Instalar dependências
# ------------------------------------------------------------
import subprocess
import sys
from pathlib import Path

projeto_path = Path.cwd()
requirements = projeto_path /  "requirements.txt"

if not requirements.exists():
    raise FileNotFoundError(f"requirements.txt não encontrado em: {requirements}")

print("✓ Instalando dependências...")

subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        str(requirements)
    ],
    check=True
)
```
